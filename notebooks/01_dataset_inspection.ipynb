```python
df_clean = df.copy()
```

```python
df_clean.shape
```

```python
df_clean['director'] = df_clean['director'].fillna('Not Specified')
```

```python
df_clean['cast'] = df_clean['cast'].fillna('Not Specified')
```

```python
df_clean['country'] = df_clean['country'].fillna('Not Specified')
```

```python
df_clean['rating'] = df_clean['rating'].fillna('Not Rated')
```

```python
df_clean.isnull().sum()
```

```python
df_clean['date_added'] = pd.to_datetime(
    df_clean['date_added'],
    errors='coerce'
)
```

```python
df_clean['date_added'] = df['date_added'].copy()
```

```python
df_clean['date_added'] = pd.to_datetime(
    df_clean['date_added'].str.strip(),
    format='mixed',
    errors='coerce'
)
```

```python
print("Original missing dates:", df['date_added'].isnull().sum())
print("Cleaned missing dates:", df_clean['date_added'].isnull().sum())
```

```python

df_clean['duration_minutes'] = (
    df_clean['duration']
    .str.extract(r'(\d+)\s*min')[0]
    .astype('Int64')
)
```

```python
df_clean['duration_seasons'] = (
    df_clean['duration']
    .str.extract(r'(\d+)\s*Season')[0]
    .astype('Int64')
)
```

```python
df_clean[['type', 'duration', 'duration_minutes', 'duration_seasons']].head(20)
```

```python
print("Movies:", (df_clean['type'] == 'Movie').sum())
print("Movies with duration in minutes:",
      df_clean.loc[df_clean['type'] == 'Movie', 'duration_minutes'].notna().sum())

print("TV Shows:", (df_clean['type'] == 'TV Show').sum())
print("TV Shows with seasons:",
      df_clean.loc[df_clean['type'] == 'TV Show', 'duration_seasons'].notna().sum())
```
